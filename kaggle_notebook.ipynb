1. Add your GitHub Personal Access Token (PAT) to **Kaggle Secrets** as `GITHUB_TOKEN`.

3. Run the cells below to clone your private repo, render the video, and push the results back.

```python
# Retrieve Secrets and Clone Repository
from kaggle_secrets import UserSecretsClient
import os

try:
    user_secrets = UserSecretsClient()
    GITHUB_TOKEN = user_secrets.get_secret("GITHUB_TOKEN")
    GITHUB_USERNAME = user_secrets.get_secret("GITHUB_USERNAME")
    GITHUB_EMAIL = user_secrets.get_secret("GITHUB_EMAIL")
    REPO_NAME = user_secrets.get_secret("REPO_NAME")
    
    os.environ['GITHUB_TOKEN'] = GITHUB_TOKEN
    os.environ['GITHUB_USERNAME'] = GITHUB_USERNAME
    os.environ['GITHUB_EMAIL'] = GITHUB_EMAIL
    os.environ['REPO_NAME'] = REPO_NAME
    
    REPO_URL = f"https://{GITHUB_USERNAME}:{GITHUB_TOKEN}@github.com/{GITHUB_USERNAME}/{REPO_NAME}.git"
    
    !git clone {REPO_URL} workspace
    %cd workspace
except Exception as e:
    print("Error retrieving secrets. Make sure GITHUB_TOKEN, GITHUB_USERNAME, GITHUB_EMAIL, and REPO_NAME are set in Kaggle Secrets.")
    print(e)
```

```python
# Render the video and push back to GitHub
!chmod +x render_and_push.sh
!./render_and_push.sh
```
